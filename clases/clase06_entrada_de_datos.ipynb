## User Input y While Loops

Hoy en día ya no se utiliza mucho, pero sirve para leer datos desde entradas, va a esperar a que el usuario escriba algo y va a devolver siempre un String ``(str)`` por eso se necesita utilizar convertidor a entero, double, ..

```python
name = input('¿Cómo te llamas?')
age = int(input('¿Cual es su edad?'))
print("Hola, " + name.upper())
print(f"su edad es {age}")
print(f'Tipo de dato de nombre: {type(name)}')
print(f'Tipo de dato de edad: {type(age)}')
```

## Estructura de control While Loop

El ciclo se repite MIENTRAS la condición sea verdadera (True), y se detiene en el instante en que la condición se vuelve falsa (False). Hay que tener cuidado con el blucle infinito por eso el bloque de código debe estar pensado para que la condición sea false.

```python
EXIT_COMMAND = "salir"

user_input = ""

while user_input.lower() != EXIT_COMMAND:
    user_input = input("Enter command: ")
    
    if user_input.lower() == EXIT_COMMAND:
        print("Saliendo del programa...")
    else:
        print(f"Ejecución: {user_input}")
```
